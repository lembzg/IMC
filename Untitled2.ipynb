```python
import pandas as pd
import numpy as np

df = pd.read_csv('/home/j39233pt/Desktop/IMC/data/round5/prices_round_5_day_2.csv', sep=';')
choc = df[df['product']=='SNACKPACK_CHOCOLATE']['mid_price'].values
van = df[df['product']=='SNACKPACK_VANILLA']['mid_price'].values

for lag in [1,2,3,5,10,20,50,100, 110, 120, 150, 160, 170, 180, 190, 200, 210, 220, 230, 240, 250, 260, 270, 300, 350, 400]:
    corr = np.corrcoef(choc[:-lag], van[lag:])[0,1]
    print(f"lag={lag}: corr={corr:.3f}")
    
```

```python
spread = choc + van
print(f"Mean: {spread.mean():.1f}")
print(f"Std: {spread.std():.1f}")
print(f"Min: {spread.min():.1f}")
print(f"Max: {spread.max():.1f}")
```

```python
# PANEL vs OXYGEN
panel = df[df['product']=='PANEL_4X4']['mid_price'].values
oxygen = df[df['product']=='OXYGEN_SHAKE_EVENING_BREATH']['mid_price'].values

print("panel leads oxygen:")
for lag in [1,2,3,5,10,20,50, 100, 120, 150, 170, 200, 300, 400, 500, 600]:
    corr = np.corrcoef(panel[:-lag], oxygen[lag:])[0,1]
    print(f"  lag={lag}: corr={corr:.3f}")

print("oxygen leads panel:")
for lag in [1,2,3,5,10,20,50, 120, 150, 170, 200, 300, 400, 500, 600]:
    corr = np.corrcoef(oxygen[:-lag], panel[lag:])[0,1]
    print(f"  lag={lag}: corr={corr:.3f}")
    
```

```python
trans = df[df['product']=='TRANSLATOR_SPACE_GRAY']['mid_price'].values
mist = df[df['product']=='TRANSLATOR_GRAPHITE_MIST']['mid_price'].values

print("translator leads mist:")
for lag in [1,2,3,5,10,20,50,100,200,300, 400, 500, 1000]:
    corr = np.corrcoef(trans[:-lag], mist[lag:])[0,1]
    print(f"  lag={lag}: corr={corr:.3f}")

print("mist leads translator:")
for lag in [1,2,3,5,10,20,50,100,200,300]:
    corr = np.corrcoef(mist[:-lag], trans[lag:])[0,1]
    print(f"  lag={lag}: corr={corr:.3f}")
```

```python
choc = df[df['product']=='SNACKPACK_CHOCOLATE']['mid_price'].values
van = df[df['product']=='SNACKPACK_VANILLA']['mid_price'].values

spread = choc + van
print(f"spread mean: {spread.mean():.1f}")
print(f"spread std: {spread.std():.1f}")
print(f"spread min: {spread.min():.1f}")
print(f"spread max: {spread.max():.1f}")

import matplotlib.pyplot as plt
plt.plot(spread)
plt.title('CHOC + VAN spread')
plt.show()
```

```python
amber = df[df['product']=='UV_VISOR_AMBER']['mid_price'].values
yellow = df[df['product']=='UV_VISOR_YELLOW']['mid_price'].values

# Check correlation
corr = np.corrcoef(amber, yellow)[0,1]
print(f"correlation: {corr:.3f}")

# Check if spread is stable (pairs trade)
spread = amber + yellow
print(f"spread mean: {spread.mean():.1f}")
print(f"spread std: {spread.std():.1f}")

# Also check simple difference
diff = amber - yellow
print(f"diff mean: {diff.mean():.1f}")
print(f"diff std: {diff.std():.1f}")
```

```python
import matplotlib.pyplot as plt
plt.plot(amber + yellow)
plt.title('AMBER + YELLOW spread')
plt.show()
```
