```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dfs = []

for day in [1, 2, 3]:
    df = pd.read_csv(f'data/prices/prices_round_4_day_{day}.csv', sep=';')
    df["day"] = day
    dfs.append(df)

prices = pd.concat(dfs, ignore_index=True)

print(prices.columns.tolist())
print(prices['product'].unique())
```

```python
hydro = prices[prices["product"] == "HYDROGEL_PACK"].copy()

print(hydro.head())
print(hydro.shape)
```

```python
hydro["mid"] = (hydro["bid_price_1"] + hydro["ask_price_1"]) / 2
hydro["spread"] = hydro["ask_price_1"] - hydro["bid_price_1"]

print(hydro[["day", "timestamp", "bid_price_1", "ask_price_1", "mid", "spread"]].head())
print(hydro["spread"].describe())
```

```python
for d in hydro["day"].unique():
    subset = hydro[hydro["day"] == d]

    plt.figure(figsize=(12, 4))
    plt.plot(subset["mid"])
    plt.title(f"HYDRO Mid Price - Day {d}")
    plt.show()
```

```python
windows = [500, 1000, 1500, 2000, 2500, 3000]
horizons = [10, 20, 50, 100]
thresholds = [1.5, 2, 2.5]

rows = []

for d in sorted(hydro["day"].unique()):
    df = hydro[hydro["day"] == d].copy()

    for window in windows:
        df["mean"] = df["mid"].rolling(window).mean()
        df["std"] = df["mid"].rolling(window).std()
        df["z"] = (df["mid"] - df["mean"]) / df["std"]

        for threshold in thresholds:
            for horizon in horizons:
                future = df["mid"].shift(-horizon) - df["mid"]

                low = df["z"] < -threshold
                high = df["z"] > threshold

                rows.append({
                    "day": d,
                    "window": window,
                    "threshold": threshold,
                    "horizon": horizon,
                    "low_count": low.sum(),
                    "high_count": high.sum(),
                    "low_avg_move": future[low].mean(),
                    "high_avg_move": future[high].mean(),
                    "low_winrate": (future[low] > 0).mean(),
                    "high_winrate": (future[high] < 0).mean(),
                })

results = pd.DataFrame(rows)
display(results)
```

```python
filtered = results[
    (results["low_count"] >= 100) &
    (results["high_count"] >= 100)
].copy()

# Score both sides together
filtered["score"] = (
    filtered["low_avg_move"] 
    - filtered["high_avg_move"]
    + (filtered["low_winrate"] - 0.5) * 10
    + (filtered["high_winrate"] - 0.5) * 10
)

display(filtered.sort_values("score", ascending=False).head(30))
```

```python
grouped = filtered.groupby(["window", "threshold", "horizon"]).agg({
    "low_avg_move": "mean",
    "high_avg_move": "mean",
    "low_winrate": "mean",
    "high_winrate": "mean",
    "low_count": "sum",
    "high_count": "sum"
}).reset_index()
grouped["score"] = (
    grouped["low_avg_move"]
    - grouped["high_avg_move"]
    + (grouped["low_winrate"] - 0.5) * 10
    + (grouped["high_winrate"] - 0.5) * 10
)
display(grouped.sort_values("score", ascending=False).head(20))
```

```python
WINDOW = 2000
THRESHOLD = 2.0

for d in sorted(hydro["day"].unique()):
    df = hydro[hydro["day"] == d].copy()

    df["mean"] = df["mid"].rolling(WINDOW).mean()
    df["std"] = df["mid"].rolling(WINDOW).std()
    df["z"] = (df["mid"] - df["mean"]) / df["std"]

    buy_signal = df["z"] < -THRESHOLD
    sell_signal = df["z"] > THRESHOLD

    plt.figure(figsize=(14, 5))
    plt.plot(df["mid"], label="mid")
    plt.plot(df["mean"], label="2000 rolling mean")

    plt.scatter(df.index[buy_signal], df.loc[buy_signal, "mid"], marker="^", label="buy signal")
    plt.scatter(df.index[sell_signal], df.loc[sell_signal, "mid"], marker="v", label="sell signal")

    plt.title(f"HYDRO Day {d}: 2000-window z-score signals")
    plt.legend()
    plt.show()
```

```python
# Extreme active-taking indicator test with spread-cost filter

windows = [100, 200, 500, 1000, 1500, 2000, 3000]
thresholds = [1.5, 2.0, 2.5, 3.0]
horizons = [200, 500, 1000, 2000]

rows = []

for d in sorted(hydro["day"].unique()):
    df = hydro[hydro["day"] == d].copy()
    df = df.reset_index(drop=True)

    for window in windows:
        df["fair"] = df["mid"].rolling(window).mean()
        df["std"] = df["mid"].rolling(window).std()
        df["z"] = (df["mid"] - df["fair"]) / df["std"]

        for threshold in thresholds:
            buy = df["z"] < -threshold
            sell = df["z"] > threshold

            for horizon in horizons:
                future_move = df["mid"].shift(-horizon) - df["mid"]

                buy_edge = future_move[buy]       # want positive
                sell_edge = -future_move[sell]    # want positive

                all_edge = pd.concat([buy_edge, sell_edge]).dropna()

                rows.append({
                    "day": d,
                    "window": window,
                    "threshold": threshold,
                    "horizon": horizon,
                    "trades": len(all_edge),
                    "avg_edge": all_edge.mean(),
                    "median_edge": all_edge.median(),
                    "winrate_gt_0": (all_edge > 0).mean(),
                    "winrate_gt_8": (all_edge > 8).mean(),
                    "winrate_gt_16": (all_edge > 16).mean(),
                })

active_results = pd.DataFrame(rows)

display(
    active_results[
        active_results["trades"] >= 30
    ].sort_values(["avg_edge", "winrate_gt_8"], ascending=False).head(30)
)
```

```python
summary = active_results[active_results["trades"] >= 30].groupby(
    ["window", "threshold", "horizon"]
).agg({
    "trades": "sum",
    "avg_edge": "mean",
    "median_edge": "mean",
    "winrate_gt_0": "mean",
    "winrate_gt_8": "mean",
    "winrate_gt_16": "mean",
}).reset_index()

display(
    summary.sort_values(["avg_edge", "winrate_gt_8"], ascending=False).head(30)
)
```
