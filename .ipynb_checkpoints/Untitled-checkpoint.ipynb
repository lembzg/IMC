```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
prices = pd.read_csv("/home/j39233pt/Desktop/IMC/data/prices/prices_round_3_day_0.csv", sep=";")
prices.head()
prices.columns
```

```python
trades = pd.read_csv("/home/j39233pt/Desktop/IMC/ROUND_3/trades_round_3_day_0.csv", sep=";")
trades.head()
trades.columns
```

```python
hydro = prices[prices["product"] == "HYDROGEL_PACK"].copy()
hydro.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(hydro["mid_price"])
plt.title("Hydro Mid Price")
plt.show()
```

```python
alpha = 0.2

ema = []
prev = hydro["mid_price"].iloc[0]

for m in hydro["mid_price"]:
    prev = alpha * m + (1 - alpha) * prev
    ema.append(prev)

hydro["ema"] = ema
```

```python
plt.figure(figsize=(12,5))
plt.plot(hydro["mid_price"], label="mid")
plt.plot(hydro["ema"], label="ema")
plt.legend()
plt.title("Mid vs EMA")
plt.show()
```

```python
hydro["signal"] = hydro["mid_price"] - hydro["ema"]

plt.figure(figsize=(12,5))
plt.plot(hydro["signal"])
plt.title("Signal (mid - ema)")
plt.show()
```

```python
trades_hydro = trades[trades["symbol"] == "HYDROGEL_PACK"].copy()

plt.figure(figsize=(12,5))

plt.plot(hydro["timestamp"], hydro["mid_price"], label="mid")

plt.scatter(
    trades_hydro["timestamp"],
    trades_hydro["price"],
    color="red",
    s=5,
    label="trades"
)

plt.legend()
plt.title("Where trades actually happen")
plt.show()
```
