```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
prices = pd.read_csv("/home/j39233pt/Desktop/IMC/data/prices/prices_round_3_day_0.csv", sep=";")
prices.head()
prices.columns
```

```python
trades = pd.read_csv("/home/j39233pt/Desktop/IMC/ROUND_3/trades_round_3_day_0.csv", sep=";")
trades.head()
trades.columns
```

```python
hydro = prices[prices["product"] == "HYDROGEL_PACK"].copy()
hydro.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(hydro["mid_price"])
plt.title("Hydro Mid Price")
plt.show()
```

```python
alpha = 0.2

ema = []
prev = hydro["mid_price"].iloc[0]

for m in hydro["mid_price"]:
    prev = alpha * m + (1 - alpha) * prev
    ema.append(prev)

hydro["ema"] = ema
```

```python
plt.figure(figsize=(12,5))
plt.plot(hydro["mid_price"], label="mid")
plt.plot(hydro["ema"], label="ema")
plt.legend()
plt.title("Mid vs EMA")
plt.show()
```

```python
hydro["signal"] = hydro["mid_price"] - hydro["ema"]

plt.figure(figsize=(12,5))
plt.plot(hydro["signal"])
plt.title("Signal (mid - ema)")
plt.show()
```

```python
trades_hydro = trades[trades["symbol"] == "HYDROGEL_PACK"].copy()

plt.figure(figsize=(12,5))

plt.plot(hydro["timestamp"], hydro["mid_price"], label="mid")

plt.scatter(
    trades_hydro["timestamp"],
    trades_hydro["price"],
    color="red",
    s=5,
    label="trades"
)

plt.legend()
plt.title("Where trades actually happen")
plt.show()
```

```python
hydro["spread"] = hydro["ask_price_1"] - hydro["bid_price_1"]
print(hydro["spread"].describe())
hydro["spread"].plot(title="Spread over time")
```

```python
hydro["returns"] = hydro["mid_price"].diff()
print("Autocorrelation lag-1:", hydro["returns"].autocorr(lag=1))
print("Autocorrelation lag-2:", hydro["returns"].autocorr(lag=2))
print("Autocorrelation lag-3:", hydro["returns"].autocorr(lag=3))
```

```python
hydro["next_return"] = hydro["mid_price"].shift(-1) - hydro["mid_price"]
corr = hydro["signal"].corr(hydro["next_return"])
print("Signal vs next tick return correlation:", corr)
```

```python
hydro["fv_ema_fast"] = hydro["mid_price"].ewm(alpha=0.3).mean()
hydro["fv_ema_slow"] = hydro["mid_price"].ewm(alpha=0.05).mean()

for col in ["fv_ema_fast", "fv_ema_slow"]:
    signal = hydro["mid_price"] - hydro[col]
    corr = signal.corr(hydro["next_return"])
    print(f"{col}: {corr:.4f}")
    
```

```python
for alpha in [0.8, 0.9, 0.95, 1.0]:
    fv = hydro["mid_price"].ewm(alpha=alpha).mean()
    signal = hydro["mid_price"] - fv
    corr = signal.corr(hydro["next_return"])
    print(f"alpha={alpha}: {corr:.4f}")
```
